```python
# Food Order System

foods = ["Rice", "Pizza", "Burger", "Chicken", "Pasta"]

order = input("Enter food item: ")

if order in foods:
    print("Order successful")
else:
    print("Item not available")
```

```python
# Salary Bonus Calculator

salary = float(input("Enter salary: ₦"))

if salary > 50000:
    bonus = salary * 0.10
elif 30000 <= salary <= 50000:
    bonus = salary * 0.05
else:
    bonus = salary * 0.02

final_salary = salary + bonus

print("Bonus: ₦", bonus)
print("Final Salary: ₦", final_salary)
```
